4. Utiliza histogramas para determinar la distribución de los valores representados en cada variable.
*   ¿Se corresponde con lo obtenido en el cálculo de asimetría? Como verás, los datos reales son más complejos que la teoría. Por esta razón, recuerda siempre acompañar el análisis de la asimetría con algún gráfico como un histograma.

**NOTA**. Para esto también puedes ocupar los gráficos `kde` ([kernel density estimation](https://www.cienciadedatos.net/documentos/pystats02-kernel-density-estimation-kde-python.html)) que crean una curva continua y suave expandiendo la idea del histograma.

```python
columnas_numericas = df.select_dtypes(include=["number"]).columns

# Configurar el tamaño del gráfico
fig, axes = plt.subplots(nrows=len(columnas_numericas), ncols=1, figsize=(6, 4 * len(columnas_numericas)))

# Si hay más de una variable, iteramos sobre cada una
for i, col in enumerate(columnas_numericas):
    sns.histplot(df[col].dropna(), bins=30, kde=True, color="#3182bd", alpha=0.6, ax=axes[i])
    axes[i].set_title(f'Distribución de {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Densidad')

plt.tight_layout()  # Ajusta los espacios entre gráficos
plt.show()
```


**¿Se corresponde con lo obtenido en el cálculo de asimetría? Como verás, los datos reales son más complejos que la teoría. Por esta razón, recuerda siempre acompañar el análisis de la asimetría con algún gráfico como un histograma.**

5. Emplea boxplots para mostrar la distribución de los datos a través de sus cuartiles.

```python
num_cols = df.select_dtypes(include=['number']).columns

fig, axes = plt.subplots(8,1, figsize=(15,15))
axes = axes.ravel()
for col, ax in zip(df[num_cols], axes):
  sns.boxplot(x=df[col], ax=ax, showmeans=True) # ---> Seaborn
  ax.set(xlabel=f'{col}')
```

*   Como podrás observar hay valores atípicos en todas las variables. Ejecuta el siguiente código para identificar los valores atípicos en la variable `person_age`

```python
percentile_25 = df["person_age"].quantile(0.25)
percentile_75 = df["person_age"].quantile(0.75)
iqr = percentile_75 - percentile_25
upper_limit = percentile_75 + 1.5 * iqr
lower_limit = percentile_25 - 1.5 * iqr
IQR_outliers = df[(df["person_age"] < lower_limit) | (df["person_age"] > upper_limit)]
IQR_outliers
```

La variable `loan_status` será la variable de salida (o a predecir en un modelo de ML). Analiza su relación con el resto de las variables a través de los siguientes gráficos:

8. Un box plot para visualizar la distribución de `loan_percent_income` según el `loan_status`. Interpreta el resultado.

```python
plt.figure(figsize=(8,6))

sns.boxplot(x='loan_status', y='loan_percent_income', data=df, showfliers=True)

# Personalización del gráfico
plt.title("Distribución de loan_percent_income según loan_status", fontsize=14)
plt.xlabel("Estado del Préstamo", fontsize=12)
plt.ylabel("Porcentaje de Ingreso del Préstamo", fontsize=12)
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)

plt.show()
```



*   loan_percent_income : representa qué porcentaje del ingreso de una persona se destina al pago del préstamo.

 loan_status:
0 → No incumplimiento (Préstamos pagados correctamente).
1 → Incumplimiento o mora (Préstamos con problemas de pago).

*   Para loan_status = 0 (No incumplimiento): La mediana es más baja, lo que indica que la mayoría de las personas que no incumplen destinan un menor porcentaje de sus ingresos al préstamo.

*   Para loan_status = 1 (Incumplimiento o mora):
 La mediana es más alta, lo que significa que las personas que tienen problemas para pagar sus préstamos generalmente destinan un mayor porcentaje de sus ingresos al pago del préstamo.

*   Los prestatarios que destinan un menor porcentaje de sus ingresos al préstamo tienen menos riesgo de incumplimiento, mientras que aquellos con un porcentaje más alto tienen mayor probabilidad de estar en mora.
